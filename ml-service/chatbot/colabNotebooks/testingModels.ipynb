```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# --------------------------
# 1️⃣  Set your model path
# --------------------------
MODEL_PATH = "Qwen/Qwen2-0.5B-Instruct"   # change this if your model folder has a different name

# --------------------------
# 2️⃣  Load model and tokenizer
# --------------------------
print("🔹 Loading your LegalMate model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_PATH,
    torch_dtype=torch.float16,
    device_map="auto"
)

# --------------------------
# 3️⃣  Start simple chat loop
# --------------------------
print("\n💬 LegalMate Chatbot is ready! (type 'exit' to quit)\n")

while True:
    query = input("👤 You: ").strip()
    if query.lower() in ["exit", "quit"]:
        print("👋 Goodbye!")
        break

    inputs = tokenizer(query, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=300,
            temperature=0.7,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # remove repeated user text from start if needed
    if response.startswith(query):
        response = response[len(query):].strip()

    print(f"🤖 LegalMate: {response}\n")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

MODEL_NAME = "Qwen/Qwen2-0.5B-Instruct"  # works with any Instruct model

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

def chat_with_qwen(prompt):
    messages = [
        {"role": "system", "content": "You are LegalMate, a helpful AI that gives legal advice."},
        {"role": "user", "content": prompt}
    ]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(text, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=256,
            temperature=0.7,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response.split("assistant")[-1].strip()

if __name__ == "__main__":
    while True:
        q = input("You: ")
        if q.lower() == "exit":
            break
        print("Bot:", chat_with_qwen(q))
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel
import torch

# -------------------------------
# 1️⃣ Define paths
# -------------------------------
BASE_MODEL = "Qwen/Qwen2-0.5B"  # You fine-tuned this one

ADAPTER_PATH = "/content/drive/MyDrive/LegalMateData/qwen2_stage4/qwen2_stage4_supremecourtsJudgments_finetuned"

# -------------------------------
# 2️⃣ Load tokenizer and base model
# -------------------------------
tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL)

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

# -------------------------------
# 3️⃣ Load your LoRA fine-tuned adapter
# -------------------------------
model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
print("✅ LegalMate fine-tuned LoRA model loaded successfully!")

# -------------------------------
# 4️⃣ Define chat function
# -------------------------------
def chat_with_qwen(prompt):
    messages = [
    {"role": "system", "content": "You are a friendly chatbot that answers normally."},
    {"role": "user", "content": prompt}
]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(text, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=256,
            temperature=0.7,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return response.split("assistant")[-1].strip()

# -------------------------------
# 5️⃣ Run interactive chat
# -------------------------------
if __name__ == "__main__":
    print("💬 LegalMate is ready! Type 'exit' to quit.")
    while True:
        q = input("You: ")
        if q.lower() == "exit":
            break
        print("Bot:", chat_with_qwen(q))
```
